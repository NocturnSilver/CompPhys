```python
import numpy as np
import math
import matplotlib.pyplot as plt
import seaborn as sns   
from matplotlib.pyplot import figure
import matplotlib.pyplot as plt
from importlib import reload
```

# defining functions 

```python
import numpy as np 

def cos_forward(h,t):
    return (np.cos(t + h)-np.cos(t))/h

def e_forward(h,t):
    return (np.exp(t + h)-np.exp(t))/h

def cos_center(h,t):
    return (np.cos(t + (h/2))-np.cos(t - (h/2)))/h

def e_center(h,t):
    return (np.exp(t + (h/2))-np.exp(t - (h/2)))/h

def cos_extra(h,t):
    return (8*(np.cos(t + (h/4)) - np.cos(t - (h/4))) - (np.cos(t + (h/2))-np.cos(t - (h/2))))/(3*h)

def e_extra(h,t):
    return (8*(np.exp(t + (h/4)) - np.exp(t - (h/4))) - (np.exp(t + (h/2))-np.exp(t - (h/2))))/(3*h)
```

# testing the forward difference algorithm

```python
harr=np.linspace(1,1E-2,100)
harr=np.append(harr,np.linspace(1E-2,1E-5,100))
harr=np.append(harr,np.linspace(1E-5,1E-8,100))
harr=np.append(harr,np.linspace(1E-8,1E-10,100))
```

# testing the forward difference algorithm

```python
def cos_rel_error():

    err_cos_for_1 = abs((cos_forward(harr,1)+np.sin(1))/np.sin(1))
    err_cos_for_01 = abs((cos_forward(harr,0.1)+np.sin(0.1))/np.sin(0.1))
    err_cos_for_100 = abs((cos_forward(harr,100)+np.sin(100))/np.sin(100))
    
    err_cos_cen_1 = abs((cos_center(harr,1)+np.sin(1))/np.sin(1))
    err_cos_cen_01 = abs((cos_center(harr,0.1)+np.sin(0.1))/np.sin(0.1))
    err_cos_cen_100 = abs((cos_center(harr,100)+np.sin(100))/np.sin(100))
    
    err_cos_ext_1 = abs((cos_extra(harr,1)+np.sin(1))/np.sin(1))
    err_cos_ext_01 = abs((cos_extra(harr,0.1)+np.sin(0.1))/np.sin(0.1))
    err_cos_ext_100 = abs((cos_extra(harr,100)+np.sin(100))/np.sin(100))
    
    x = harr
    y1,z1,w1 = err_cos_for_1, err_cos_cen_1, err_cos_ext_1
    y01,z01,w01 = err_cos_for_01, err_cos_cen_01, err_cos_ext_01
    y100,z100,w100 = err_cos_for_100, err_cos_cen_100, err_cos_ext_100
    
    fig, (ax1, ax2, ax3) = plt.subplots(3,figsize=(15,15))
    
    ax1.loglog(x,y1, label = 't = 1')
    ax1.loglog(x,y01, label = 't = 0.1')
    ax1.loglog(x,y100, label = 't = 100')
    ax1.set_title('forward difference algorithm')
    ax1.grid()
    ax1.legend()
    ax1.set_xlabel('h')
    ax1.set_ylabel('$\epsilon\_approx$')
    
    ax2.loglog(x,z1, label = 't = 1')
    ax2.loglog(x,z01, label = 't = 0.1')
    ax2.loglog(x,z100, label = 't = 100')
    ax2.set_title('central difference algorithm')
    ax2.grid()
    ax2.legend()
    ax2.set_xlabel('h')
    ax2.set_ylabel('$\epsilon\_approx$')
    
    ax3.loglog(x,w1, label = 't = 1')
    ax3.loglog(x,w01, label = 't = 0.1')
    ax3.loglog(x,w100, label = 't = 100')
    ax3.set_title('extrapolated difference algorithm')
    ax3.grid()
    ax3.legend()
    ax3.set_xlabel('h')
    ax3.set_ylabel('$\epsilon\_approx$')
    

    
    
cos_rel_error()
```
