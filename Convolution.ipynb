### c.) Plot the output 𝑦[𝑛] by using and making a Python, Matlab, or Sci-lab program for the convolution sum

$$𝑦[𝑛]=𝑥[𝑛]∗𝐼[𝑛]= \sum_{n=-\infty}^{\infty} x[k]𝐼[𝑛−𝑘]$$

```python
n, y, SUM, k = np.arange(-20,20,1), [], [], 0

for i in n:
    if i>=0:
        while k<=i:
            a = 0.5**k
            SUM.append(a)
            k = k + 1
        total = sum(SUM)
        y.append(total)
    else:
        y.append(0)
        
fig, ax1 = plt.subplots(1, figsize=(10,5))        
ax1.stem(n,y)
ax1.grid()
```

### a.) Find an expression for the input and Impulse signal in terms of a Kronecker-delta function.


$$𝑥[𝑛]= \sum_{k=1}^{4} \delta[n−k] = \delta[n−1] + \delta[n−2] + \delta[n−3] + \delta[n−4]$$

$$ 
x[n] =
\begin{cases} 
      0 & n < 1 \\
      1 & 1 \leq n \leq 4\\
      0 & n > 4 \\
   \end{cases}
$$

$$I[𝑛]= \sum_{k=0}^{2} \delta[n-k] = \delta[n−0]+ \delta[n−1] + \delta[n−2]$$

$$ 
I[n] =
\begin{cases} 
      0 & n < 0 \\
      1 & 0 \leq n \leq 2\\
      0 & n > 2 \\
   \end{cases}
$$

$$y[n] = x[n]*I[n] = \sum_{k=-\infty}^{\infty} x[k]\delta[n−k]$$

$$x[n]*I[n] = x[n] * \{ \delta[n−0]+ \delta[n−1] + \delta[n−2] \}$$
\\
$$x[n]*I[n] = x[n] *\delta[n−0]+ x[n] *\delta[n−1] + x[n] *\delta[n−2]$$

$$x[n]*I[n] = x[n]+ x[n-1] + x[n-2]$$

\\

$$x[n]*I[n] = \delta[n−1] + \delta[n−2] + \delta[n−3] + \delta[n−4]$$

$$+\delta[n−2] + \delta[n−3] + \delta[n−4] + \delta[n−5] + \delta[n−3] + \delta[n−4] + \delta[n−5] + \delta[n−6]$$

$$x[n]*I[n] = 1\delta[n−1] + 2\delta[n−2] + 3\delta[n−3] + 3\delta[n−4]+ 2\delta[n−5] + 1\delta[n−6]$$

$$x[n]*I[n] = \{1,2,3,3,2,1 \}$$

### b.) Create a program for the convolution sum then plot the output y[n].

```python
def ddelta5(a,x):
    y = np.zeros(len(a))
    y[11+x:10+x+5] = 1 
    return y

def ddelta6(a,x):
    y = np.zeros(len(a))
    y[10+x:10+x+3] = 1 
    return y
    
def convo(a,x):
    y = np.zeros(len(a))
    y[11+x:10+x+7] = 1 
    y[12+x:10+x+6] = 2 
    y[13+x:10+x+5] = 3 
    return y
    
t = np.arange(-10,10,1)
l1 = ddelta5(t,0)
l2 = ddelta6(t,0)
l3 = convo(t,0)

fig, (ax1,ax2,ax3) = plt.subplots(3,figsize=(10,10))
ax1.stem(t, l1)
ax1.set_title('x[n]')
ax1.grid()

ax2.stem(t, l2)
ax2.set_title('i[n]')
ax2.grid()

ax3.stem(t, l3)
ax3.set_title('x[n]i[n]')
ax3.grid()
```

# 5.) If the input signal 𝑥[𝑛]=sin𝜋n/6 is periodic and the impulse response is 𝐼[𝑛]=0.5^𝑛 𝐻[𝑛].

### a.) Plot the value of the output signal y[n] by convolution sum using a program.


```python
n = np.arange(-100,100,1)

def sine(n):
    m = np.sin(np.pi*n/6)
    return(m)
def x_n(n):
    if n>=0:
        return(1)
    else:
        return(0)
def I_n(n):
    y = (0.5**n)*x_n(n)
    return(y)

n = np.arange(-20,20,1)
y = []
for i in n:
    a = -1000
    b = 1000
    conv = []
    while a<=b:
        f = sine(a)*I_n(i-a)
        conv.append(f)
        a = a + 1
    SUM = sum(conv)
    y.append(SUM)

fig, ax1 = plt.subplots(1,figsize=(10,5))
ax1.stem(n,y)
ax1.grid()
  
```
