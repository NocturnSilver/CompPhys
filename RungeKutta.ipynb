# Part 2. Standard Form of the Simple Harmonic ODE (5 pts.)

Recall that the differential equation for the simple harmonic motion is:

$$\frac{d^{2}x}{dt^{2}} = -\frac{k}{m} x$$

We can write this ODE into the form

$$\frac{dy}{dt} = f(t,y)$$

where

$$y = \begin{pmatrix}x(t)\\ v(t)\end{pmatrix} \\ f(t,x(t),v(t)) = \begin{pmatrix} v(t)\\ -\frac{k}{m} x \end{pmatrix}$$

1. write a function in python that will return the value of the vector f(t,y)

```python
# processes values and gives out the velocity and force
array1 = [0,0]

def f(t,y):
    array1[0] = y[1]
    array1[1] = -(k/m)*y[0]
    array = np.array(array1)
    return array
```

# Part 3. Solving the Differential Equation using RK2 (10 pts.)

Create a program that will implement the RK2 method to solve the ODE of the simple harmonic motion 
numerically. Recall that in RK2 method, the values of the position and velocity are determined by the 
algorithm:

$$𝒚_{𝒏 + 𝟏} = 𝒚_𝒏 + 𝒌_𝟐 \\
𝒌_𝟐 = ℎ𝒇(𝑡_𝑛 + \frac{ℎ}{2}, 𝒚_𝒏 + \frac{𝒌𝟏}{2})\\
𝒌_𝟏 = ℎ𝒇(𝑡_𝑛,𝒚_𝒏)
$$

(Note: Symbols in bold implies that the quantity is a vector)

```python
y = [x0, v0]
t = 0
pos, time, vel  =  [], [], []

i = 0

while i < 2000:
    
    time.append(t)
    pos.append(y[0])
    vel.append(y[1])
    
    k1 = h*f(t, y)
    k2 = h*f(t + (h/2), y + (k1/2))
    
    print(k1, k2)
    
    # update next value of y
    # we get position and velocity form an 
    # array y that only asks for x0 and v0
    
    y += k2   
    
    t += h
    i += 1

```

# Part 4. Solving the Differential Equation using RK4 (10 pts.)

Create a program that will implement the RK4 method to solve the ODE of the simple harmonic motion  numerically. Recall that in RK4 method, the values of the position and velocity are determined by the algorithm:

$$
𝒚_{𝒏 + 𝟏} = 𝒚_𝒏 + \frac{1}{6}(𝒌_𝟏 + 𝟐𝒌_𝟐 + 𝟐𝒌_𝟑 + 𝒌_𝟒) \\
𝒌_𝟏 = ℎ𝒇(𝑡_𝑛,𝒚_𝒏) \\
𝒌_𝟐 = ℎ𝒇(𝑡_𝑛 + \frac{ℎ}{2} ,𝒚_𝒏 + \frac{𝒌_𝟏}{2} ) \\
𝒌_𝟑 = ℎ𝒇(𝑡_𝑛 + \frac{ℎ}{2} ,𝒚_𝒏 + \frac{𝒌_𝟐}{2} ) \\
𝒌_𝟒 = ℎ𝒇(𝑡_𝑛 + ℎ,𝒚_𝒏 + 𝒌_𝟑)
$$

1-2.) Repeat numbers 1 and 2 in part 3.

```python
y = [x0, v0]
t = 0
pos_, time_, vel_  =  [], [], []

i = 0

while i < 2000:
    
    time_.append(t)
    pos_.append(y[0])
    vel_.append(y[1])
    
    k1 = h*f(t, y)
    k2 = h*f(t + (h/2), y + (k1/2))
    k3 = h*f(t + h/2,y + k2/2)
    k4 = h*f(t + h,y + k3)
    
    
    # update next value of y
    # we get position and velocity form an 
    # array y that only asks for x0 and v0
    
    y += (1/6)*(k1+(2*k2)+(2*k3)+k4)   
    
    t += h
    i += 1
```

```python
fig, (ax1, ax2) = plt.subplots(2,figsize=(10,10))

ax1.plot(time_, vel_, color = "violet")
ax1.grid()
ax1.legend()
ax1.set_xlabel('time')
ax1.set_ylabel('velocity')

ax2.plot(time_, pos_, color = "lime")
ax2.grid()
ax2.legend()
ax2.set_xlabel('time')
ax2.set_ylabel('position')
```

# Part 5. Error Assessment via Conservation of Energy (10 pts.)

1. Determine the potential, kinetic, and total energies using the following equations"
$$
𝑉 = \frac{1}{2} 𝑘𝑥(𝑡)^2             \text{(Potential Energy)} \\
𝐾 = \frac{1}{2} 𝑚𝑣(𝑡)^2            \text{(Kinetic Energy)}\\
𝐸 = 𝐾 + 𝑉                          \text{(Total Energy)} \\
$$

```python
V, K, E = [], [], []
V_, K_, E_ = [], [], []

for i in range(len(pos)):

    potential, kinetic,  = (1/2)*(k)*(pos[i]**2), (1/2)*(m)*(vel[i]**2)
    potential_, kinetic_ = (1/2)*(k)*(pos_[i]**2), (1/2)*(m)*(vel_[i]**2)
    total, total_  = kinetic + potential, kinetic_ + potential_
    
    V.append(potential) 
    K.append(kinetic) 
    E.append(total) 
    V_.append(potential_) 
    K_.append(kinetic_) 
    E_.append(total_) 
```

2. Plot the potential, kinetic, and total energy against time. Base on the graph can you say that the 
energy is conserved?

3. Plot the logarithm of the relative error:

$$𝜀 = log_{10} \begin{vmatrix}\frac{E(t) - E(0)}{E(0)}\end{vmatrix}$$

against time from the results obtained in RK2 and RK4 methods. Which one of the algorithms has a 
lesser order of magnitude of error?

```python
err_arr, err_arr_ = [], []


for i in range(len(E)):
    rel_err  = np.log10(abs((E[i]-E[0])/E[0]))
    rel_err_ = np.log10(abs((E_[i]-E_[0])/E_[0]))
    err_arr.append(rel_err)
    err_arr_.append(rel_err_)




fig, ax1 = plt.subplots(figsize=(20,10))

# velocity graphs

ax1.plot(time, err_arr, color = "violet", label = 'rk2')
ax1.plot(time_, err_arr_, color = "cyan", label = 'rk4')
ax1.grid()
ax1.legend()
ax1.set_xlabel('time')
ax1.set_ylabel('velocity')
```
