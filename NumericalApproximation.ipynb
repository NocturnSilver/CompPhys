## 3. Determine the parameter $\alpha$ and $\beta$ in the error analysis:

$$ \epsilon_{approx} = A(N)=A(2N)\approx \frac{\alpha}{N^{\beta}_c}$$


for number 2. Here 𝐴(𝑁) is the algorithmic approximation for N terms/iterations. 𝛽 is the negative of the slope in the log-log plot obtained in number 2 (assume a linear curve). Obtain the value of 𝛼 using equation (3) 

```python
def parameters():

    x = []
    
    array = cos_error()
    array = [abs(k) for k in array]
    for i in range(len(array)):
        x.append(i)
         
    n_count = np.array(x)    
    err = np.array(array)
       
    def func(b,Nc,a):
        return (-b*Nc)+a
 
    #define x and y for the curve
    xf=np.log(n_count[1:7])
    yf=np.log(abs(err[1:7]))
    
    #curve fitting
    popt,_=curve_fit(func,xf,yf)
    bt,al=popt
    alpha=10**al
    print('β =', -bt,'\n𝛼 =',alpha)
    
    return (alpha,-bt)
    
    
parameters()
```
